# 회귀 연습 문제

Linnerud 운동 데이터를 사용해서 회귀 모델을 한 단계씩 연습합니다.


## Step 0. 데이터 로드 함수

아래 함수는 Linnerud 데이터를 불러와서 입력 데이터와 정답 데이터를 나누어 반환합니다.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.core.display import display_latex

from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, root_mean_squared_error


def load_linnerud_data():
    linnerud = load_linnerud(as_frame=True)

    exercise_df = linnerud.data.copy()
    body_df = linnerud.target.copy()

    return exercise_df, body_df
```

```python
exercise_df, body_df = load_linnerud_data()

display(exercise_df.head())
display(body_df.head())
```

```python
# • 문제 1
#   운동 기록으로 Weight를 예측하려고 한다.
X = exercise_df
y = body_df['Weight']

print(X.shape)
print(y.shape)
```

```python
# 문제 2
#
# 학습 데이터와 평가 데이터를 나눠라.
#
# 조건:
# - train_test_split을 사용한다.
# - 평가 데이터 비율은 전체의 25%로 둔다.
# - 결과가 항상 같게 나오도록 random_state는 42로 둔다.
# - 나눠야 하는 대상은 X와 y다.
# - 결과 변수는 학습용 입력, 평가용 입력, 학습용 정답, 평가용 정답 4개가 필요하다.
# 힌트:
#
# - 함수 이름은 train_test_split
# - 괄호 안에는 X, y, test_size, random_state가 들어간다.
# - 왼쪽에는 4개의 변수가 와야 한다.

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# 문제 3
#
# 나눈 데이터의 크기를 확인해라.
#
# 조건:
#
# - 학습용 입력 데이터 크기 확인
# - 평가용 입력 데이터 크기 확인
# - 학습용 정답 데이터 크기 확인
# - 평가용 정답 데이터 크기 확인
# - .shape를 사용한다.
#
# 힌트:
#
# - 확인해야 할 변수는 4개다.
# - 이름은 방금 만든 X_train, X_test, y_train, y_test다.
# - 출력 결과는 각각 행과 열 개수 또는 데이터 개수를 보여줘야 한다.
print("X_train.shape: ", X_train.shape)
print("X_test.shape: ", X_test.shape)
print("y_train.shape: ", y_train.shape)
print("y_test.shape: ", y_test.shape)
```

```python
# 문제 4
#
# 입력 데이터 X_train, X_test를 스케일링해라.
#
# 조건:
#
# - StandardScaler를 사용한다.
# - 학습 데이터에는 기준을 만들면서 변환한다.
# - 평가 데이터에는 학습 데이터에서 만든 기준만 사용해서 변환한다.
# - 결과 변수는 스케일링된 학습 입력, 스케일링된 평가 입력으로 만든다.
#
# 힌트:
#
# - 먼저 스케일러 객체를 하나 만든다.
# - y_train, y_test는 스케일링하지 않는다.

scaler = StandardScaler()

# - 학습 데이터에는 fit과 transform을 같이 하는 메서드를 쓴다.
X_train_scaled = scaler.fit_transform(X_train)
# - 평가 데이터에는 transform만 쓴다.
X_test_scaled = scaler.transform(X_test)
```

```python
# 문제 5
#
# 선형 회귀 모델을 하나 만들어서 학습시켜라.
#
# 조건:
#
# - LinearRegression을 사용한다.
# - 모델 객체를 하나 만든다.
# - 스케일링된 학습 입력 데이터로 학습한다.
# - 정답 데이터는 y_train을 사용한다.
#
# 힌트:
#
# - 모델을 만드는 단계가 먼저다.
lin_reg = LinearRegression()
# - 그다음 학습 메서드를 호출한다.
# - 입력은 X_train_scaled
# - 정답은 y_train
lin_reg.fit(X_train_scaled, y_train)
```

```python
# 문제 6
#
# 학습된 모델로 예측값을 만들어라.
#
# 조건:
#
# - 학습 데이터에 대한 예측값을 만든다.
# - 평가 데이터에 대한 예측값을 만든다.
# - 학습 데이터 예측 결과는 따로 변수에 저장한다.
# - 평가 데이터 예측 결과도 따로 변수에 저장한다.

# 힌트:
#
# - 예측할 때는 학습된 모델의 예측 메서드를 사용한다.
# - 학습 입력은 X_train_scaled
y_pred_train = lin_reg.predict(X_train_scaled)
# - 평가 입력은 X_test_scaled
y_pred_test = lin_reg.predict(X_test_scaled)
# - 변수 이름은 네가 정해도 된다.
```

```python
# 문제 7
#
# 모델 성능을 숫자로 확인해라.
#
# 조건:
#
# - 학습 데이터의 R2 점수를 구한다.
# - 평가 데이터의 R2 점수를 구한다.
# - 평가 데이터의 RMSE를 구한다.
# - 각각 출력한다.
#
# 힌트:
#
# - R2는 모델 객체의 점수 계산 메서드로 구할 수 있다.
# - 학습 R2에는 X_train_scaled, y_train이 필요하다.
train_R2 = r2_score(y_train, y_pred_train)
# - 평가 R2에는 X_test_scaled, y_test가 필요하다.
test_R2 = r2_score(y_test, y_pred_test)
# - RMSE는 실제값과 예측값을 비교해서 구한다.
train_RMSE = root_mean_squared_error(y_train, y_pred_train)
# - 평가 RMSE에는 y_test와 y_pred_test가 필요하다.
test_RMSE = root_mean_squared_error(y_test, y_pred_test)
```

```python
# 문제 8
# 방금 만든 성능 값을 출력해라.

print('train_R2:', train_R2)
print('test_R2: ', test_R2)
print('train_RMSE: ', train_RMSE)
print('test_RMSE: ', test_RMSE)
```

```python
# 문제 9
#
# 이번에는 Ridge 회귀 모델을 만들어서 학습시켜라.
#
# 조건:
#
# - Ridge를 사용한다.
# - alpha 값은 1.0으로 둔다.
# - 모델 객체를 하나 만든다.
# - 스케일링된 학습 입력 데이터로 학습한다.
# - 정답 데이터는 y_train을 사용한다.
#
# 힌트:
#
# - 선형 회귀 모델을 만들었던 방식과 거의 같다.
ridge_reg = Ridge(alpha=1.0)
ridge_reg.fit(X_train_scaled, y_train)
# - 모델 이름만 LinearRegression에서 Ridge로 바뀐다.
# - Ridge는 괄호 안에 규제 강도 alpha를 넣을 수 있다.
```

```python
# • 문제 10
#
# Ridge 모델의 예측값을 만들어라.
#
# 조건:
#
# - 학습 데이터에 대한 예측값을 만든다.
# - 평가 데이터에 대한 예측값을 만든다.
# - 각각 다른 변수에 저장한다.
# - 예측에는 방금 만든 Ridge 모델을 사용한다.
#
# 힌트:
#
# - 사용할 모델: ridge_reg
ridge_reg.fit(X_train_scaled, y_train)
# - 학습 입력: X_train_scaled
# - 평가 입력: X_test_scaled
# - LinearRegression에서 예측했던 방식과 같다.
```

```python
# 문제 10
#
# Ridge 모델로 예측값을 만들어라.
#
# 조건:
#
# - 학습 데이터 예측값 1개
# - 평가 데이터 예측값 1개
# - 총 2개의 변수를 만든다.
# - 사용할 모델은 ridge_reg다.
#
# 힌트:
#
# - 학습할 때는 fit
# - 예측할 때는 다른 메서드를 쓴다.
# - LinearRegression에서 y_pred_train, y_pred_test를 만들었던 방식과 같다.
y_pred_train_ridge = ridge_reg.predict(X_train_scaled)
y_pred_test_ridge = ridge_reg.predict(X_test_scaled)
```

```python
# 문제 11
#
# Ridge 모델의 성능을 계산해라.
#
# 조건:
#
# - 학습 데이터의 R2를 계산한다.
train_R2_ridge = r2_score(y_train, y_pred_train_ridge)
# - 평가 데이터의 R2를 계산한다.
test_R2_ridge = r2_score(y_test, y_pred_test_ridge)
# - 학습 데이터의 RMSE를 계산한다.
train_RMSE_ridge = root_mean_squared_error(y_train, y_pred_train_ridge)
# - 평가 데이터의 RMSE를 계산한다.
test_RMSE_ridge = root_mean_squared_error(y_test, y_pred_test_ridge)
# - 결과는 LinearRegression 때와 헷갈리지 않게 Ridge 전용 변수명으로 저장한다.


# 힌트:
#
# - 학습 실제값: y_train
# - 학습 예측값: Ridge의 학습 예측값
# - 평가 실제값: y_test
# - 평가 예측값: Ridge의 평가 예측값
# - LinearRegression에서 성능 계산했던 방식과 같다.
```

```python
# 문제 12
# LinearRegression과 Ridge의 성능을 비교하기 쉽게 표로 만들어라.
#
# 조건:
#
# - pandas의 DataFrame을 사용한다.
# - 행은 2개다.
#   - LinearRegression
#   - Ridge
#
# - 컬럼은 아래 5개를 포함한다.
#   - 모델 이름
#   - 학습 R2
#   - 평가 R2
#   - 학습 RMSE
#   - 평가 RMSE
#
# - 마지막에 표를 출력한다.
#
# 힌트:
#
# - 먼저 모델별 성능 정보를 리스트 형태로 모은다.
results = [
    {
        'model': 'LinearRegression',
        'train_R2': train_R2,
        'test_R2': test_R2,
        'train_RMSE': train_RMSE,
        'test_RMSE': test_RMSE
    },
    {
        'model': 'Ridge',
        'train_R2': train_R2_ridge,
        'test_R2': test_R2_ridge,
        'train_RMSE': train_RMSE_ridge,
        'test_RMSE': test_RMSE_ridge
    }
]
# - 각 모델의 성능은 딕셔너리 형태로 넣으면 편하다.
model_dic = {
    'LinearRegression': lin_reg,
    'Ridge': ridge_reg,
}
# - LinearRegression에서 만든 성능 변수와 Ridge에서 만든 성능 변수를 함께 사용한다.

results_df = pd.DataFrame(results)
display(results_df)
```

```python
# 문제 13
# 성능 비교표에서 평가 RMSE가 가장 낮은 모델 이름을 찾아라.
best_model_name = results_df.loc[results_df['test_RMSE'].idxmin()]['model']
best_model = model_dic[best_model_name]
```

```python
# 문제 14
# 새로운 운동 기록 2개를 만들어서, 가장 좋은 모델로 Weight를 예측해라.
new_exercise = pd.DataFrame({
    'Chins': [8, 15],
    'Situps': [90, 120],
    'Jumps': [40, 55]
})

new_test_scaled = scaler.transform(new_exercise)

# best_model.fit(new_train_scaled, y_train)
#
y_new_pred = best_model.predict(new_test_scaled)
print(y_new_pred)

best_y_pred = best_model.predict(X_test_scaled)

compare_df = pd.DataFrame({
    'actual_Weight': y_test,
    'predicted_Weight': best_y_pred
})

display(compare_df)
```
